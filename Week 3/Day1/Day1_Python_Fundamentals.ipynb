## 1.Variables 

```python
name = "Asma" 
age = 22 
is_student = True 
height = 5.4
print(name, age, is_student, height)
```

## 2.Data types

```python
print(type(name), type(age), type(is_student), type(height))
```

## 3.Lists

```python
fruits = ["apple", "banana", "mango"]
fruits.append("orange")
print(fruits)
print(fruits[0])
```

## 4. Dictionaries

```python
student = {"name": "Asma", "course": "IT", "gpa": 3.1}
print(student["name"])
student["city"] = "Lahore" 
```

```python
print(student)
print(student.keys())
print(student.values())
```

## 5. Conditions

```python
marks = 85
if marks >= 90: 
    print("A grade") 
elif marks >= 75: 
    print("B grade") 
else: print("C grade")
```

## 6. Loop

```python
for fruit in fruits:
    print(fruit)

for i in range(1, 6):
    print("Number:", i)
```

## 9. Functions

```python
def greet(name):
    return f"Hello, {name}! Welcome to Data Analytics."

print(greet("Asma"))

def calculate_average(numbers):
    return sum(numbers) / len(numbers)

print(calculate_average([80, 90, 70, 85]))
```

## 10. List Comprehensions

```python
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]
print(squares)

even_numbers = [n for n in numbers if n % 2 == 0]
print(even_numbers)
```

## 11. Modules

```python
import math
print(math.sqrt(16))
print(math.pi)

import random
print(random.randint(1, 100))
```

## 12. Exception Handling

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

try:
    num = int("abc")
except ValueError:
    print("Error: Invalid number format!")
finally:
    print("This always runs, error ho ya na ho.")
```
