## 1. NumPy Arrays

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr)
print(type(arr))
print(arr.shape)
```

## 2. Array Operations & Vectorization

```python
arr2 = np.array([1, 2, 3, 4, 5])
print(arr2 + 10)
```

## 3. Indexing

```python
print(arr2[0])
```

```python
print(arr2[-1])
```

```python
print(arr2[1:4])
```

```python
print(arr2[arr2 > 2])
```

## 4. Pandas Series

```python
import pandas as pd
prices = pd.Series([80000, 1500, 3000, 20000],
index=['Laptop', 'Mouse', 'Keyboard', 'Monitor'])
print(prices)
print(prices['Laptop'])
```

## 5. Pandas DataFrames

```python
data = { 'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
'price': [80000, 1500, 3000, 20000],
'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics'] }
df = pd.DataFrame(data)
print(df)
df.head()
```
